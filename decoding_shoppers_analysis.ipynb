```python
#loaded the data set using pandas
import pandas as pd

df = pd.read_csv(r"C:\Users\vinay\Downloads\customer_shopping_behavior.csv")  
```

```python
#checked if any null values or missing data is present in the dataset
df.isnull().sum()
```

```python
#fillig the missing values in the review rating column with median rating of the product category
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
#rechecking if there are any null values again
df.isnull().sum()
```

```python
#renamed the columns to snace casing for uniformity

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#created a new column called age_group so that its in numeric format

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(15)
```

```python
# created new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
#dropped the column promo_code_used as it has same data as the discount_applied
df = df.drop('promo_code_used', axis=1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Outlier Detection on Purchase Amount - IQR method
Q1 = df['purchase_amount'].quantile(0.25)
Q3 = df['purchase_amount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['purchase_amount'] < lower_bound) | (df['purchase_amount'] > upper_bound)]
print(f"Number of outliers in purchase_amount: {outliers.shape[0]}")
```

```python
#visualised it using boxplot

plt.figure(figsize=(8,4))
sns.boxplot(x=df['purchase_amount'])
plt.title('Purchase Amount - Outlier Check')
plt.show()
```

```python
#Correlation Heatmap between numeric variables

numeric_cols = ['age', 'purchase_amount', 'review_rating', 'previous_purchases']
corr = df[numeric_cols].corr()

plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Between Numeric Features')
plt.show()
```
